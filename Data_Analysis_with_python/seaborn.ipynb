Seaborn is a Python library used for statistical data visualization.

It is built on top of Matplotlib and helps us create attractive, informative graphs with less code

```python
import seaborn as sns
```

```python
## Basic plotting with seaborn
tips = sns.load_dataset('tips')
tips
```

```python
## Create a scatter plot 

import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of total Bill vs Tip")
plt.show()
```

```python
### Line plot 

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot of Total Bill by size")
plt.show()
```

```python
## Categorical plots
# Bar plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot of Total By Day")
plt.show
```

```python
## Box Plot 
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
## Distribution of the particular data
## violin plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
## Histograms 
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## KDE Plot 

sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## pair plot 
sns.pairplot(tips)
```

```python
## Heat map --- relatioship between one feature to another feature
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
